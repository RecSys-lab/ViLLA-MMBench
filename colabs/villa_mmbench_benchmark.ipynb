## 🚀 **[Step 2] Use the Framework**

**I. Load Configurations**

```python
import os
from villa_mmbench.utils.utils import readConfigs

config = readConfigs("config.yml")
print(f"Starting '{config['general']['name']}' ... Configurations loaded successfully!")
display(config)

# [Hint] You can change the configurations in ViLLA-MMBench/villa_mmbench/config.yml
config["general"]["root_path"] = "ViLLA-MMBench"
```

**II. Prepare MovieLens and split into train and test sets**

```python
from villa_mmbench.data.movielens import prepareML

train_df, test_df, genre_dict = prepareML(config)

# Logs
print("\nTrain DataFrame:\n", train_df.head(5))
print("\nTest DataFrame:\n", test_df.head(5))
# print("\nGenre Dictionary:\n", genre_dict)
```

**III. Load text, visual, and audio embeddings**

```python
from villa_mmbench.data.modalities import prepareModalities

# Load different modalities
train_df, test_df, train_set, modalities_dict = prepareModalities(config,
                                                                  train_df,
                                                                  test_df)

#
print("\nTrain DataFrame:\n", train_df.head(5))
print("\nTest DataFrame:\n", test_df.head(5))
print("\nTrain Set:\n", train_set)
print("\nModalities Dictionary:\n", modalities_dict)
```

**IV. Hyperparameter tuning using grid search**

```python
from villa_mmbench.process.grid import gridSearch

final_models = gridSearch(config, train_df, train_set, modalities_dict)
```

**V. Generate recommendations for the test set**

```python
from villa_mmbench.process.processes import generateLists

generateLists(config, train_df, train_set, test_df, genre_dict, final_models)
```
